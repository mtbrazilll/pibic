```python
import os
import re

def contar_pontos_em_arquivos(diretorio):
    """
    Conta os pontos em cada arquivo .pcd no diretório especificado.
    Cada linha do arquivo é considerada como um ponto.
    """
    resultados = {}
    
    # Listar todos os arquivos no diretório
    for arquivo in os.listdir(diretorio):
        # Filtrar apenas os arquivos .pcd
        if arquivo.endswith('.pcd'):
            caminho_arquivo = os.path.join(diretorio, arquivo)
            
            try:
                with open(caminho_arquivo, 'r') as f:
                    # Contar o número de linhas no arquivo
                    linhas = f.readlines()
                    num_pontos = len(linhas)
                    
                    # Extrair o nome base da instância sem extensão
                    nome_base = arquivo.replace('.pcd', '')
                    
                    # Armazenar resultado
                    resultados[nome_base] = num_pontos
                    
                    print(f"Arquivo: {arquivo}, Pontos: {num_pontos}")
                    
                    # Verificar se o número no nome do arquivo corresponde ao número de pontos
                    num_nome = re.search(r'\d+', nome_base)
                    if num_nome:
                        num_nome = int(num_nome.group())
                        if num_nome != num_pontos:
                            print(f"  Aviso: O número no nome ({num_nome}) é diferente do número de pontos ({num_pontos})")
                    
            except Exception as e:
                print(f"Erro ao processar o arquivo {arquivo}: {e}")
    
    return resultados

# Diretório contendo os arquivos
diretorio = "../instancias/tsp"

# Contar os pontos em cada arquivo
print("=== Contagem de Pontos nas Instâncias ===")
resultados = contar_pontos_em_arquivos(diretorio)

# Resumo
print("\n=== Resumo ===")
print(f"Total de arquivos .pcd processados: {len(resultados)}")

# Ordenar por número de pontos
ordenados = sorted(resultados.items(), key=lambda x: x[1])
print("\nInstâncias em ordem crescente de pontos:")
for nome, pontos in ordenados:
    print(f"{nome}: {pontos} pontos")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Circle

# Pontos (iguais para os 3 gráficos)
pontos = np.array([[2, 3], [1.5, 2.5], [2.5, 2], [1.8, 1.5]])

# Função para plotar
def plotar_cobertura(pontos, centros=[], raio=1, titulo=''):
    fig, ax = plt.subplots()
    ax.scatter(pontos[:,0], pontos[:,1], color='blue')
    
    for centro in centros:
        circulo = Circle(centro, raio, fill=False, color='black')
        ax.add_patch(circulo)
    
    ax.set_xlim(-1, 6)
    ax.set_ylim(-1, 6)
    ax.set_aspect('equal')
    ax.set_title(titulo)
    plt.show()

# Exemplo 1: 4 discos
centros_4 = [[2,3], [1.5,2.5], [2.5,2], [1.8,1.5]]
plotar_cobertura(pontos, centros_4, raio=1.2, titulo='Quantidade de discos: 4')

# Exemplo 2: 1 disco
centros_1 = [[2,2.25]]
plotar_cobertura(pontos, centros_1, raio=1.2, titulo='Quantidade de discos: 1')

# Exemplo 3: sem discos
plotar_cobertura(pontos, centros=[], titulo='')
```

```python
import math

# Pontos originais
points = {
    "p2": (-0.3, -0.3),
    "d1": (0.0, 1.0),  # Centro do disco d1
    "d3": (1.5, 2.5),  # Centro do círculo pontilhado d3 (originalmente b1)
    "d4": (-1.5, -0.5) # Centro do círculo pontilhado d4 (originalmente b2)
}

# Ângulo de rotação em graus (horário é negativo)
angle_degrees = -30

# Converter ângulo para radianos
angle_radians = math.radians(angle_degrees)

# Calcular seno e cosseno
cos_angle = math.cos(angle_radians)
sin_angle = math.sin(angle_radians)

# Dicionário para armazenar os pontos rotacionados
rotated_points = {}

print("Calculando rotação de 30 graus no sentido horário:")

for name, (x, y) in points.items():
    x_new = x * cos_angle - y * sin_angle
    y_new = x * sin_angle + y * cos_angle
    rotated_points[name] = (x_new, y_new)
    print(f"Ponto original {name}: ({x:.2f}, {y:.2f}) -> Ponto rotacionado {name}': ({x_new:.2f}, {y_new:.2f})")

# --- Geração do Código TikZ ---

# Pegar as coordenadas rotacionadas formatadas
p2_rot = rotated_points['p2']
d1_rot = rotated_points['d1']
d3_center_rot = rotated_points['d3'] # Novo centro de d3
d4_center_rot = rotated_points['d4'] # Novo centro de d4

tikz_code = f"""
\\begin{{figure}}[h!]
    \\centering
    \\begin{{tikzpicture}}[scale=0.6]
        % --- Definição dos pontos NÃO rotacionados ---
        \\coordinate (p1) at (0.5,1.7);
        \\coordinate (p3) at (8.6,0.3);
        \\coordinate (p4) at (7.3,1.8);
        \\coordinate (d2) at (8.0,1); % Centro de d2
        \\coordinate (b3) at (9.5,-0.5); % Centro de d5
        \\coordinate (b4) at (6.5,2.5);  % Centro de d6

        % --- Definição dos pontos ROTACIONADOS ---
        \\coordinate (p2) at ({p2_rot[0]:.2f},{p2_rot[1]:.2f}); % p2 rotacionado
        \\coordinate (d1) at ({d1_rot[0]:.2f},{d1_rot[1]:.2f}); % d1 (centro) rotacionado
        \\coordinate (d3_center) at ({d3_center_rot[0]:.2f},{d3_center_rot[1]:.2f}); % centro de d3 rotacionado
        \\coordinate (d4_center) at ({d4_center_rot[0]:.2f},{d4_center_rot[1]:.2f}); % centro de d4 rotacionado

        % --- Desenho dos elementos NÃO rotacionados ---
        % Disco d2 (Grupo da Direita)
        \\draw[thick, fill=gray!50] (d2) circle (1.5);
        \\node at (d2) {{$d_2$}};
        % Círculos pontilhados d5 e d6
        \\draw[dashed] (b3) circle (1.5);
        \\node at (b3) {{$d_5$}};
        \\draw[dashed] (b4) circle (1.5);
        \\node at (b4) {{$d_6$}};
        % Pontos p1, p3, p4
        \\foreach \\pt/\\name/\\pos in {{p1/$p_1$/above right, p3/$p_3$/above right, p4/$p_4$/above right}}{{
            \\fill (\\pt) circle (2pt);
            \\node[\\pos] at (\\pt) {{\\name}};
        }}

        % --- Desenho dos elementos ROTACIONADOS ---
        % Disco d1 (Grupo Inferior)
        \\draw[thick, fill=gray!50] (d1) circle (1.5);
        \\node at (d1) {{$d_1$}};
        % Círculos pontilhados d3 e d4
        \\draw[dashed] (d3_center) circle (1.5);
        \\node at (d3_center) {{$d_3$}};
        \\draw[dashed] (d4_center) circle (1.5);
        \\node at (d4_center) {{$d_4$}};
        % Ponto p2
        \\fill (p2) circle (2pt);
        \\node[below left] at (p2) {{$p_2$}}; % Ajustar posição se necessário

    \\end{{tikzpicture}}
    \\caption{{\\textit{{CMSA}} ao \\SiglaP (com d4, p2, d1, d3 rotacionados 30 graus horário)}}
    \\label{{fig:exemplo_cmsa_rotacionado}}
\\end{{figure}}
"""

print("\n--- Código TikZ Gerado ---")
print(tikz_code)

# Para fins de visualização, salvar o código TikZ em um arquivo
# with open("rotated_figure_corrected.tex", "w") as f:
#     f.write("\\documentclass{standalone}\n")
#     f.write("\\usepackage{tikz}\n")
#     f.write("\\begin{document}\n")
#     f.write(tikz_code)
#     f.write("\\end{document}\n")
# print("\nCódigo TikZ também salvo em 'rotated_figure_corrected.tex'")
```
